## Group Project: Building Your AI Chess Bot ##

So far, we have learnt:
- Different basic types of AI-based search algorithms for winning in a chess game such as Combinational search, Minimax algorithm, Alpha-beta pruning, Negamax algorithm, SSS* algorthm, etc.
- Implementation of various basic AI chess games such as Tic-Tac-Toe, Connect-FOUR and Hexapawn. 

In the Group Project, form a group to design and implement an AI-based Chess Bot with:
- At least 3 levels of AI challenges, one above the others
- Build both interactive H2M and M2M - non-interactive AI Chess Game to demonstrate Chess Bot in various levels of challenge. 
- Highly recommend to add your own AI algorithm to beat the classical methods
- Extra marks will be given with nice looking interface.

## Project Term Paper (Technical Report) ##

As part of the project deliverable, each project group have to submit a Project Term Paper to discuss the literature review of AI-game/AI-Chess bot, related technology and their proposed AI Chess Bot with experimental results/ simulation results. 

The TOC (Table of Content) include:
- Project Title
- Abstract (200 - 250 words)
- Keywords (4 - 6)
- Introduction
- Literature Review
- Methodology
- Implementation
- Experimental Results
- Performance Analysis
- Future Work
- Conclusion
- Reference 

Note:
1. Total no. of page: 8 - 12 pages, double-column, single-spacing
2. Paper format: Please refer to the attached IEEE Paper Template in the directory folder. 
3. Similarity: Less than 20%.
